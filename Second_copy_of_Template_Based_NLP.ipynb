```python
import random
import pandas as pd

class MedicalNLG:
    def __init__(self):
        self.templates = {
            'clinical': [
                "Patient presents with {symptoms}. Diagnosis: {disease}.",
                "Clinical assessment shows {symptoms}. This indicates {disease}.",
                "Based on symptoms {symptoms}, the patient has {disease}."
            ],
            'simple': [
                "If you have {symptoms}, you may have {disease}.",
                "{symptoms} are signs of {disease}."
            ],
            'detailed': [
                "The patient reports the following symptoms: {symptoms}. "
                "After medical evaluation, the diagnosis is {disease}. "
                "Please consult a healthcare provider."
            ]
        }

    def format_symptoms(self, symptom_list, max_symptoms=5):
        if len(symptom_list) == 0:
            return "no symptoms"
        symptoms = symptom_list[:max_symptoms]
        if len(symptoms) == 1:
            return symptoms[0]
        elif len(symptoms) == 2:
            return f"{symptoms[0]} and {symptoms[1]}"
        else:
            return ", ".join(symptoms[:-1]) + f", and {symptoms[-1]}"

    def generate(self, disease, symptoms, template_type='clinical'):
        symptoms_text = self.format_symptoms(symptoms)
        disease_name = disease.title()
        template = random.choice(self.templates[template_type])
        return template.format(disease=disease_name, symptoms=symptoms_text)

# Test the system
nlg = MedicalNLG()
print("✓ NLG System Created Successfully!")
```

```python
# Interactive Symptom Checker with Fluency Evaluation
import pandas as pd
import random

def evaluate_fluency(text):
    """Evaluate the fluency of generated text (score out of 10)"""
    score = 0

    # Check 1: Capitalization (2 points)
    if text and text[0].isupper():
        score += 2

    # Check 2: Proper punctuation (2 points)
    if text and text[-1] in '.!?':
        score += 2

    # Check 3: Sentence length (2 points)
    words = text.split()
    if 10 <= len(words) <= 30:
        score += 2
    elif len(words) > 5:
        score += 1

    # Check 4: Contains medical keywords (2 points)
    medical_keywords = ['symptom', 'doctor', 'medication', 'consult', 'diagnosis', 'treatment', 'health']
    if any(keyword in text.lower() for keyword in medical_keywords):
        score += 2

    # Check 5: No excessive punctuation (2 points)
    if '...' not in text and '!!' not in text and '??' not in text:
        score += 2

    return min(10, score)

class SymptomChecker:
    def __init__(self):
        self.df = None
        self.symptom_columns = []
        self.all_symptoms = []

    def load_data(self):
        """Load the dataset"""
        print("Loading medical database...")
        self.df = pd.read_csv('data.csv')
        self.symptom_columns = [col for col in self.df.columns if col != 'diseases']
        # Clean symptom names (replace underscores with spaces)
        self.all_symptoms = [col.replace('_', ' ') for col in self.symptom_columns]
        print(f"✓ Loaded {len(self.df)} patient records")
        print(f"✓ Database contains {len(self.all_symptoms)} possible symptoms")
        return True

    def match_symptoms_to_disease(self, user_symptoms):
        """Find disease that matches the given symptoms"""
        best_match = None
        best_score = 0
        matched_symptoms = []

        for idx, row in self.df.iterrows():
            disease = row['diseases']
            # Count how many user symptoms match this disease
            score = 0
            disease_symptoms = []

            for symptom in user_symptoms:
                # Convert symptom to column format (spaces to underscores)
                symptom_col = symptom.lower().replace(' ', '_')
                if symptom_col in self.symptom_columns:
                    if row[symptom_col] == 1:
                        score += 1
                        disease_symptoms.append(symptom)

            if score > best_score and score > 0:
                best_score = score
                best_match = disease
                matched_symptoms = disease_symptoms

        return best_match, best_score, matched_symptoms

    def get_response(self, symptoms_list):
        """Generate appropriate response based on number of symptoms"""
        # Remove empty strings and clean
        symptoms_list = [s.strip().lower() for s in symptoms_list if s.strip()]

        if len(symptoms_list) == 0:
            return {
                'type': 'no_symptoms',
                'message': "You didn't enter any symptoms. Please consult a doctor if you're feeling unwell.",
                'advice': "Rest and stay hydrated."
            }

        elif len(symptoms_list) == 1:
            # Single symptom - minor issue
            responses = [
                f"📍 You reported: {symptoms_list[0]}\n\n✅ This appears to be a minor symptom. Take rest and over-the-counter medication if needed.\n💊 Recommendation: Paracetamol or Ibuprofen as needed.\n⏰ It should resolve within 2-3 days.\n⚠️ Consult a doctor if it persists beyond 3 days.",

                f"Symptom detected: {symptoms_list[0]}\n\n🔹 This is likely not a serious health issue.\n💊 Take appropriate medication and rest.\n🥤 Stay hydrated and monitor your condition.\n📅 Follow up with a doctor only if symptoms worsen.",

                f"✓ Symptom: {symptoms_list[0]}\n\n🟢 SEVERITY: Mild\n💊 Action: Take tablet (Paracetamol 500mg) as needed\n🛌 Rest is recommended\n📞 No immediate medical consultation needed\n✨ Should resolve with self-care"
            ]
            message = random.choice(responses)
            return {
                'type': 'minor',
                'symptom': symptoms_list[0],
                'message': message,
                'advice': "Self-care with rest and medication",
                'fluency_score': evaluate_fluency(message)
            }

        else:
            # Multiple symptoms - find matching disease
            disease, score, matched_symptoms = self.match_symptoms_to_disease(symptoms_list)

            if disease and score >= 2:
                # Found matching disease
                disease_name = disease.replace('_', ' ').title()

                # Generate response using template
                templates = [
                    f"⚠️ MULTIPLE SYMPTOMS DETECTED: {', '.join(matched_symptoms[:5])}\n\n🏥 Based on your symptoms, you may have: {disease_name}\n\n📋 Match confidence: {score}/{len(symptoms_list)} symptoms matched\n\n👨‍⚕️ RECOMMENDATION: Please consult a healthcare provider for proper diagnosis and treatment.\n🚫 Do not self-medicate.",

                    f"🔴 ALERT: Multiple symptoms reported!\n\nSymptoms: {', '.join(matched_symptoms[:5])}\n\n🎯 Possible condition: {disease_name}\n\n⚠️ This requires medical attention.\n📞 Please schedule an appointment with a doctor.\n💊 Do not take any medication without prescription.",

                    f"🏥 MEDICAL CONSULTATION RECOMMENDED\n\nYour symptoms ({', '.join(matched_symptoms[:5])}) suggest: {disease_name}\n\n📊 Symptom match: {score} out of {len(symptoms_list)}\n\n👩‍⚕️ Action Required: Visit a healthcare provider for:\n   - Proper diagnosis\n   - Prescribed medication\n   - Treatment plan\n\n🚨 If symptoms are severe, seek emergency care."
                ]
                message = random.choice(templates)
                return {
                    'type': 'serious',
                    'disease': disease_name,
                    'matched_symptoms': matched_symptoms,
                    'score': score,
                    'total_symptoms': len(symptoms_list),
                    'message': message,
                    'advice': "Consult a doctor immediately",
                    'fluency_score': evaluate_fluency(message)
                }
            else:
                # Multiple symptoms but no clear match
                message = f"⚠️ You reported {len(symptoms_list)} symptoms: {', '.join(symptoms_list)}\n\n❓ No specific disease match found in our database.\n\n👨‍⚕️ Since you have multiple symptoms, we recommend consulting a healthcare provider for proper evaluation.\n📝 Keep track of your symptoms and when they started."
                return {
                    'type': 'unclear',
                    'symptoms': symptoms_list,
                    'message': message,
                    'advice': "Medical consultation recommended",
                    'fluency_score': evaluate_fluency(message)
                }

    def suggest_symptoms(self):
        """Show common symptoms to help user"""
        common = ['fever', 'cough', 'headache', 'fatigue', 'nausea',
                  'chest pain', 'shortness of breath', 'dizziness',
                  'sore throat', 'muscle pain']
        print("\n📋 Common symptoms you can enter:")
        print(", ".join(common[:10]))
        print("(Or type any symptom you're experiencing)")

def main():
    print("="*60)
    print("🏥 SYMPTOM CHECKER & MEDICAL ADVISORY SYSTEM")
    print("="*60)
    print("\nThis system analyzes your symptoms and provides recommendations.")
    print("-"*60)

    # Initialize system
    checker = SymptomChecker()

    try:
        checker.load_data()
    except FileNotFoundError:
        print("\n❌ Dataset not found! Please ensure 'data.csv' is in the current directory.")
        return

    print("\n✅ System ready!")

    while True:
        print("\n" + "="*60)
        print("MAIN MENU")
        print("="*60)
        print("1. Check my symptoms")
        print("2. View common symptoms list")
        print("3. Exit")

        choice = input("\nEnter your choice (1-3): ").strip()

        if choice == '1':
            print("\n" + "-"*40)
            print("SYMPTOM CHECKER")
            print("-"*40)

            checker.suggest_symptoms()

            print("\n👉 Enter your symptoms (separate multiple symptoms with commas):")
            print("Example: fever, cough, headache")

            symptoms_input = input("\nYour symptoms: ").strip()

            # Parse symptoms
            symptoms = [s.strip().lower() for s in symptoms_input.split(',')]

            print("\n" + "="*50)
            print("ANALYZING YOUR SYMPTOMS...")
            print("="*50)

            # Get response
            response = checker.get_response(symptoms)

            # Display message
            print(f"\n{response['message']}")

            # Display fluency score
            print(f"\n📊 FLUENCY SCORE: {response['fluency_score']}/10")

            print("\n" + "-"*40)
            print(f"📌 MEDICAL ADVICE: {response['advice']}")
            print("-"*40)

            if response['type'] == 'minor':
                print("\n💡 TIP: Get plenty of rest and stay hydrated.")
            elif response['type'] == 'serious':
                print("\n🚨 IMPORTANT: This is not a definitive diagnosis.")
                print("   Always consult a qualified healthcare professional.")

        elif choice == '2':
            checker.suggest_symptoms()

        elif choice == '3':
            print("\n👋 Thank you for using the Symptom Checker!")
            print("⚠️ Remember: This tool is for informational purposes only.")
            print("   Always consult a doctor for medical advice.")
            break

        else:
            print("❌ Invalid choice. Please enter 1, 2, or 3.")

if __name__ == "__main__":
    main()
```

```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models

# Load MNIST Dataset
(X_train, y_train), (X_test, y_test) = datasets.mnist.load_data()

# Normalize Images
X_train = X_train / 255.0
X_test = X_test / 255.0

# Reshape Images
X_train = X_train.reshape(-1,28,28,1)
X_test = X_test.reshape(-1,28,28,1)

# Build CNN Model
model = models.Sequential([
    layers.Conv2D(32, (3,3), activation='relu',
                  input_shape=(28,28,1)),
    layers.MaxPooling2D((2,2)),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

# Compile Model
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Train Model
model.fit(X_train, y_train, epochs=5)

# Evaluate Model
loss, accuracy = model.evaluate(X_test, y_test)

print("Test Accuracy:", accuracy)

# Predict First Test Image
prediction = model.predict(X_test[:1])

print("Predicted Digit:", prediction.argmax())
print("Actual Digit:", y_test[0])
```
